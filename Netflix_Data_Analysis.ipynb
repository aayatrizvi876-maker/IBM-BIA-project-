## 5. Dataset Description
The dataset contains 600 synthetic records with fields: `show_id`, `title`, `type`, `director`, `country`, `date_added`, `release_year`, `rating`, `duration`, and `listed_in`. It has been bundled with the project so the workflow can run offline.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("netflix_sample_dataset.csv")
df.head()
```

```python
print("Rows and columns:", df.shape)
df.info()
df.isna().sum().sort_values(ascending=False)
```

## 6. Data Cleaning
Duplicate rows are removed. Missing director, country, and rating values are assigned the label `Not Specified`; release year is converted to numeric and date-added values are parsed as dates. These choices preserve records while making missingness visible.

```python
df = df.drop_duplicates()
for col in ["director", "country", "rating"]:
    df[col] = df[col].fillna("Not Specified")
df["release_year"] = pd.to_numeric(df["release_year"], errors="coerce")
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
print(df.isna().sum())
```

## 7. Descriptive Analysis
Frequency tables describe content type, countries, genres, and ratings. The `describe()` method summarizes the numerical release-year variable.

```python
print(df["type"].value_counts())
print("\\nRelease-year summary:")
print(df["release_year"].describe())
print("\\nTop countries:")
print(df["country"].value_counts().head(10))
print("\\nGenre frequencies:")
print(df["listed_in"].value_counts())
print("\\nRatings by type:")
pd.crosstab(df["type"], df["rating"])
```

## 8. Visualization — Content Type
A bar chart compares the number of Movie and TV Show records. Bar charts are suitable for comparing category frequencies.

```python
df["type"].value_counts().plot(kind="bar", title="Titles by Content Type")
plt.xlabel("Content Type"); plt.ylabel("Count"); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```

## 9. Visualization — Release Year
A line chart shows the number of titles by release year, making temporal distribution easier to inspect.

```python
df["release_year"].value_counts().sort_index().plot(kind="line", figsize=(10,4), marker="o")
plt.title("Titles by Release Year"); plt.xlabel("Release Year"); plt.ylabel("Count"); plt.tight_layout(); plt.show()
```

## 10. Visualization — Countries and Genres
Horizontal bar charts compare category frequencies. In real catalog data, a row may contain multiple countries or genres; those fields should be split before counting individual country/genre contributions.

```python
df["country"].value_counts().head(8).sort_values().plot(kind="barh", figsize=(8,4))
plt.title("Top Countries"); plt.xlabel("Count"); plt.tight_layout(); plt.show()

df["listed_in"].value_counts().sort_values().plot(kind="barh", figsize=(8,4))
plt.title("Genre Distribution"); plt.xlabel("Count"); plt.tight_layout(); plt.show()
```

## 11. Visualization — Ratings
A stacked bar chart compares rating frequencies for Movies and TV Shows. It provides a descriptive comparison without implying causation.

```python
pd.crosstab(df["type"], df["rating"]).plot(kind="bar", stacked=True, figsize=(9,4))
plt.title("Ratings by Content Type"); plt.xlabel("Content Type"); plt.ylabel("Count"); plt.tight_layout(); plt.show()
```
